```python
from browser_use import Agent, Browser

browser = Browser(
    executable_path=r"C:\Program Files\Google\Chrome\Application\chrome.exe",
    user_data_dir=r"D:\BroserUse\User Data",
    profile_directory="Profile 1",
    keep_alive=True
)


async def run_task(instruction: str):
    agent = Agent(
        task=instruction,
        llm=llm,
        browser=browser,
    )

    history = await agent.run()

    return history


async def main():
    await browser.start()

    try:
        while True:
            instruction = input("\nYou: ")

            if instruction.lower() in ["exit", "quit"]:
                break

            result = await run_task(instruction)
            print("\nAgent:", result)

    finally:
        await browser.stop()


if __name__ == "__main__":
    import asyncio
    await main()
```
